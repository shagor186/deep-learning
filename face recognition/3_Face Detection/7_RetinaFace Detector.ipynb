# 1. Basic

## 1. Individual Image Detection

```python
from retinaface import RetinaFace
import cv2

# Image path
img_path = "../dataset/test/ronaldo.jpg"
img = cv2.imread(img_path)

# Face detection
faces = RetinaFace.detect_faces(img_path)

# Draw bounding boxes & landmarks
for key in faces:
    identity = faces[key]
    x1, y1, x2, y2 = identity['facial_area']
    cv2.rectangle(img, (x1, y1), (x2, y2), (0, 255, 0), 2)

    for point in identity['landmarks'].values():
        cv2.circle(img, point, 2, (0, 0, 255), -1)

cv2.imshow("RetinaFace - Image", img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

## 2. Video Detection (MP4, AVI)

```python
import cv2
from retinaface import RetinaFace

video_path = "../dataset/vid/s.mp4"
cap = cv2.VideoCapture(video_path)

while True:
    ret, frame = cap.read()
    if not ret:
        break

    faces = RetinaFace.detect_faces(frame)

    for key in faces:
        identity = faces[key]
        x1, y1, x2, y2 = identity['facial_area']
        cv2.rectangle(frame, (x1, y1), (x2, y2), (0, 255, 0), 2)

        for point in identity['landmarks'].values():
            cv2.circle(frame, point, 2, (0, 0, 255), -1)

    cv2.imshow("RetinaFace - Video", frame)
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```

## 3. Webcam Detection (Real-time)

```python
import cv2
from retinaface import RetinaFace

cap = cv2.VideoCapture(0)  # Webcam

while True:
    ret, frame = cap.read()
    if not ret:
        break

    faces = RetinaFace.detect_faces(frame)

    for key in faces:
        identity = faces[key]
        x1, y1, x2, y2 = identity['facial_area']
        cv2.rectangle(frame, (x1, y1), (x2, y2), (0, 255, 0), 2)

        for point in identity['landmarks'].values():
            cv2.circle(frame, point, 2, (0, 0, 255), -1)

    cv2.imshow("RetinaFace - Webcam", frame)
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```
