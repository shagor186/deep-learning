## 1. Single Image Face Detection

```python
from ultralytics import YOLO
import cv2

# Step 1: Model লোড
model = YOLO('yolov8n-face.pt')

# Step 2: Image লোড
img_path = "../dataset/test/ronaldo.jpg"
results = model(img_path, show=True)  # show=True থাকলে উইন্ডোতে ফলাফল দেখাবে

# Step 3: Bounding Box ড্র করা এবং সংরক্ষণ
for result in results:
    boxes = result.boxes.xyxy  # [x1, y1, x2, y2]
    confs = result.boxes.conf  # confidence
    print(f"Detected Faces: {len(boxes)}")

# Save result image
results[0].save(filename="output_face_detected.jpg")
```

## 2. Video Face Detection

```python
from ultralytics import YOLO
import cv2

model = YOLO('yolov8n-face.pt')

video_path = "../dataset/vid/s.mp4"
cap = cv2.VideoCapture(video_path)

while True:
    ret, frame = cap.read()
    if not ret:
        break
    
    results = model.predict(source=frame, show=True)
    
    # Optional: result save করতে চাইলে নিচের কোড ব্যবহার করুন
    # results[0].save(filename="frame_output.jpg")
    
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```

## 3. Webcam Face Detection

```python
from ultralytics import YOLO
import cv2

model = YOLO('yolov8n-face.pt')

cap = cv2.VideoCapture(0)  # 0 মানে default webcam

while True:
    ret, frame = cap.read()
    if not ret:
        break

    results = model.predict(source=frame, show=True)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```
