# 1. Basic

## 1. Individual Image Face Detection

```python
from mtcnn import MTCNN
import cv2

# Image load
img = cv2.imread("../dataset/test/ronaldo.jpg")
rgb_img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Initialize MTCNN detector
detector = MTCNN()

# Detect faces
faces = detector.detect_faces(rgb_img)

# Draw bounding boxes and landmarks
for face in faces:
    x, y, w, h = face['box']
    cv2.rectangle(img, (x, y), (x+w, y+h), (0, 255, 0), 2)
    
    # Draw landmarks
    for key, point in face['keypoints'].items():
        cv2.circle(img, point, 3, (0, 0, 255), -1)

cv2.imshow("MTCNN Image Face Detection", img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

## 2. Video File Face Detection

```python
from mtcnn import MTCNN
import cv2

# Video path
video_path = "../dataset/vid/s.mp4"
cap = cv2.VideoCapture(video_path)

detector = MTCNN()

while True:
    ret, frame = cap.read()
    if not ret:
        break

    rgb_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
    faces = detector.detect_faces(rgb_frame)

    for face in faces:
        x, y, w, h = face['box']
        cv2.rectangle(frame, (x, y), (x+w, y+h), (0, 255, 0), 2)
        for key, point in face['keypoints'].items():
            cv2.circle(frame, point, 3, (0, 0, 255), -1)

    cv2.imshow("MTCNN Video Face Detection", frame)
    
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```

## 3. Webcam Face Detection

```python
from mtcnn import MTCNN
import cv2

cap = cv2.VideoCapture(0)  # Webcam
detector = MTCNN()

while True:
    ret, frame = cap.read()
    if not ret:
        break

    rgb_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
    faces = detector.detect_faces(rgb_frame)

    for face in faces:
        x, y, w, h = face['box']
        cv2.rectangle(frame, (x, y), (x+w, y+h), (0, 255, 0), 2)
        for key, point in face['keypoints'].items():
            cv2.circle(frame, point, 3, (0, 0, 255), -1)

    cv2.imshow("MTCNN Webcam Face Detection", frame)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```
