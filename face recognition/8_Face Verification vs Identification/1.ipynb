```python
from sklearn.neighbors import KNeighborsClassifier
import numpy as np

# Example: suppose you have 3 embeddings (128-D) for 3 people
v1 = np.random.rand(128)  # Alice's embedding
v2 = np.random.rand(128)  # Bob's embedding
v3 = np.random.rand(128)  # Charlie's embedding

# Training embeddings and labels
X_train = np.array([v1, v2, v3])
y_train = ["Alice", "Bob", "Charlie"]

# Train KNN classifier
knn = KNeighborsClassifier(n_neighbors=3, metric='euclidean')
knn.fit(X_train, y_train)

# Predict new face
# Suppose embedder.embeddings([new_face]) returns a 128-D vector
v_new = np.random.rand(128)  # Replace with actual embedding from your model
predicted_label = knn.predict([v_new])
print("Predicted Identity:", predicted_label[0])
```

```python
import os
import numpy as np
from tensorflow.keras.preprocessing.image import load_img, img_to_array
from tensorflow.keras.applications.mobilenet_v2 import MobileNetV2, preprocess_input
from sklearn.neighbors import KNeighborsClassifier

# -----------------------------
# 1️⃣ Load Pretrained Model (Embedding Extractor)
# -----------------------------
embedder = MobileNetV2(weights='imagenet', include_top=False, pooling='avg', input_shape=(224,224,3))

# -----------------------------
# 2️⃣ Load Images & Labels
# Folder structure:
# dataset/
#   train/
#     Alice/
#       1.jpg, 2.jpg...
#     Bob/
#       1.jpg, 2.jpg...
# -----------------------------
train_dir = "img_dataset/train"
X_train = []
y_train = []

for person_name in os.listdir(train_dir):
    person_path = os.path.join(train_dir, person_name)
    if not os.path.isdir(person_path):
        continue
    for img_name in os.listdir(person_path):
        img_path = os.path.join(person_path, img_name)
        img = load_img(img_path, target_size=(224,224))
        img_array = img_to_array(img)
        img_array = np.expand_dims(img_array, axis=0)
        img_array = preprocess_input(img_array)
        
        # Extract embedding
        embedding = embedder.predict(img_array)[0]
        X_train.append(embedding)
        y_train.append(person_name)

X_train = np.array(X_train)
y_train = np.array(y_train)

# -----------------------------
# 3️⃣ Train KNN
# -----------------------------
knn = KNeighborsClassifier(n_neighbors=3, metric='euclidean')
knn.fit(X_train, y_train)

# -----------------------------
# 4️⃣ Predict New Image
# -----------------------------
def predict_face(image_path):
    img = load_img(image_path, target_size=(224,224))
    img_array = img_to_array(img)
    img_array = np.expand_dims(img_array, axis=0)
    img_array = preprocess_input(img_array)
    
    embedding = embedder.predict(img_array)[0]
    prediction = knn.predict([embedding])
    return prediction[0]

# Example usage:
new_face_path = "img_dataset/test/15.jpg"
print("Predicted Identity:", predict_face(new_face_path))
```
