```python
# Example 1: Binary Classification
from tensorflow.keras import layers, models

model = models.Sequential([
    layers.Flatten(input_shape=(28,28)),   # ইমেজ ফ্ল্যাট করা
    layers.Dense(128, activation='relu'),  # Hidden layer
    layers.Dense(1, activation='sigmoid')  # Output layer (Binary)
])

model.summary()
```

```python
# Example 2: Multi-Class Classification
model = models.Sequential([
    layers.Flatten(input_shape=(28,28)),   # ইনপুট
    layers.Dense(128, activation='relu'),  # Hidden layer
    layers.Dense(10, activation='softmax') # 10-ক্লাস আউটপুট
])
model.summary()
```

```python
# Example 3: Regression Output
model = models.Sequential([
    layers.Dense(64, activation='relu', input_shape=(10,)),
    layers.Dense(1, activation='linear')  # কোন activation নেই, সরাসরি মান
])
model.summary()
```

```python
# Example 4: Softmax Shape Check
import tensorflow as tf
x = tf.constant([[1.0, 2.0, 3.0]])
y = tf.nn.softmax(x)  # Softmax ফাংশন প্রয়োগ
print(y.numpy())
```

```python
# Example 5: Sigmoid Function
import tensorflow as tf
z = tf.constant([0.0, 2.0, -1.0])
y = tf.nn.sigmoid(z)
print(y.numpy())
```

```python
# Example 6: Model for 3-Class Classification
model = models.Sequential([
    layers.Dense(32, activation='relu', input_shape=(64,)),
    layers.Dense(3, activation='softmax')  # তিনটি আউটপুট ক্লাস
])
model.summary()
```

```python
# Example 7: Linear Output Example
import tensorflow as tf
x = tf.constant([[5.0]])
y = tf.keras.activations.linear(x)
print(y.numpy())  # আউটপুট = ইনপুট (কোন পরিবর্তন নয়)
```

```python
# Example 8: Output Probability Check
x = tf.constant([[1.0, 2.0, 3.0]])
prob = tf.nn.softmax(x)
print("Sum:", tf.reduce_sum(prob).numpy())  # সর্বদা 1.0 হয়
```

```python
# Example 9: Custom Dense Output
from tensorflow.keras import layers
dense = layers.Dense(4, activation='softmax')
out = dense(tf.random.normal([1,8]))
print(out)
```

```python
# Example 10: Full CNN with Output Layer
cnn = models.Sequential([
    layers.Conv2D(32, (3,3), activation='relu', input_shape=(64,64,3)),
    layers.MaxPooling2D((2,2)),
    layers.Flatten(),
    layers.Dense(64, activation='relu'),
    layers.Dense(5, activation='softmax')  # 5-ক্লাস আউটপুট
])
cnn.summary()
```
