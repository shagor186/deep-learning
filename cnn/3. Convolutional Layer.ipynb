```python
from tensorflow.keras import layers, models

# CNN মডেল তৈরি করা হচ্ছে
model = models.Sequential([
    layers.Input(shape=(32, 32, 3)),         # ইনপুট লেয়ার (RGB ইমেজ)
    layers.Conv2D(32, (3,3), activation='relu', padding='valid'), # ৩x৩ ফিল্টার, ৩২টি কনভোলিউশন
    layers.MaxPooling2D(pool_size=(2,2)),    # পুলিং লেয়ার (স্পেশাল সাইজ হ্রাস)
    layers.Conv2D(64, (3,3), activation='relu', padding='same'),  # আরেকটি কনভোলিউশন
    layers.MaxPooling2D(pool_size=(2,2)),    # আরেকটি পুলিং লেয়ার
    layers.Flatten(),                        # ফিচারগুলো একত্র করা
    layers.Dense(128, activation='relu'),    # ফুলি কানেক্টেড লেয়ার
    layers.Dense(10, activation='softmax')   # আউটপুট লেয়ার (১০ ক্লাস)
])

model.summary()
```

```python
from tensorflow.keras import layers, models
# keras থেকে লেয়ার ও মডেল ইমপোর্ট করা হচ্ছে

model = models.Sequential([
    layers.Input(shape=(32, 32, 3)),          # ইনপুট ইমেজ: 32x32 RGB
    layers.Conv2D(32, (3,3), activation='relu'), # ৩২টি ৩x৩ ফিল্টার প্রয়োগ (feature extract)
    layers.MaxPooling2D((2,2)),               # ২x২ পুলিং (স্পেশাল ডাইমেনশন কমানো)
    layers.Conv2D(64, (3,3), activation='relu'), # ৬৪টি ফিল্টার (উচ্চ স্তরের ফিচার)
    layers.MaxPooling2D((2,2)),               # আরও ডাউনস্যাম্পল করা হচ্ছে
    layers.Flatten(),                         # সব ফিচার একত্র করা হচ্ছে
    layers.Dense(128, activation='relu'),     # ফুলি কানেক্টেড লেয়ার
    layers.Dense(10, activation='softmax')    # আউটপুট: ১০ ক্লাসে শ্রেণীবিন্যাস
])
```

```python
# Code Example 1: Basic Conv2D 
from tensorflow.keras import layers, models

model = models.Sequential([
    layers.Input(shape=(28,28,1)),            # ইনপুট 28x28 গ্রেস্কেল
    layers.Conv2D(16, (3,3), activation='relu') # ৩x৩ ফিল্টার, ১৬টি ফিচার
])
model.summary()
```

```python
# Code Example 2: Conv2D + Padding
model = models.Sequential([
    layers.Input(shape=(32,32,3)),              # RGB ইমেজ
    layers.Conv2D(32, (3,3), activation='relu', padding='same') # Padding same
])
model.summary()
```

```python
# Code Example 3: Conv2D + MaxPooling
model = models.Sequential([
    layers.Input(shape=(32,32,3)),
    layers.Conv2D(32, (3,3), activation='relu'),
    layers.MaxPooling2D(pool_size=(2,2))       # ২x২ ম্যাক্সপুলিং
])
model.summary()
```

```python
# Code Example 4: Multi-layer Conv
model = models.Sequential([
    layers.Input(shape=(64,64,3)),
    layers.Conv2D(32, (3,3), activation='relu'),
    layers.Conv2D(64, (3,3), activation='relu'),
    layers.MaxPooling2D(2,2)
])
model.summary()
```

```python
# Code Example 5: Conv + Flatten + Dense
model = models.Sequential([
    layers.Input(shape=(28,28,1)),
    layers.Conv2D(16, (3,3), activation='relu'),
    layers.Flatten(),
    layers.Dense(10, activation='softmax')    # আউটপুট 10 ক্লাস
])
model.summary()
```

```python
# Code Example 6: Conv with Stride
model = models.Sequential([
    layers.Input(shape=(32,32,3)),
    layers.Conv2D(32, (3,3), strides=(2,2), activation='relu')
])
model.summary()
```

```python
# Code Example 7: Conv + ReLU Activation
model = models.Sequential([
    layers.Input(shape=(28,28,1)),
    layers.Conv2D(16, (3,3)),
    layers.Activation('relu')                  # আলাদা ReLU লেয়ার
])
model.summary()
```

```python
# Code Example 8: Conv + Batch Normalization
model = models.Sequential([
    layers.Input(shape=(32,32,3)),
    layers.Conv2D(32, (3,3), activation='relu'),
    layers.BatchNormalization()                # ব্যাচ নরমালাইজেশন
])
model.summary()
```

```python
# Code Example 9: Conv + Dropout
model = models.Sequential([
    layers.Input(shape=(32,32,3)),
    layers.Conv2D(32, (3,3), activation='relu'),
    layers.Dropout(0.5)                        # ৫০% নিউরন ড্রপ
])
model.summary()
```

```python
# Code Example 10: Full CNN Example
model = models.Sequential([
    layers.Input(shape=(32,32,3)),
    layers.Conv2D(32, (3,3), activation='relu'),
    layers.MaxPooling2D(2,2),
    layers.Conv2D(64, (3,3), activation='relu'),
    layers.MaxPooling2D(2,2),
    layers.Flatten(),
    layers.Dense(128, activation='relu'),
    layers.Dense(10, activation='softmax')    # আউটপুট 10 ক্লাস
])
model.summary()
model.summary()
```

# 4. Math Solution

```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D

# 🔹 Input Image (5x5) — reshape to (height, width, channels)
# channels=1 কারণ grayscale image
I = np.array([
    [1, 2, 3, 0, 1],
    [0, 1, 2, 3, 1],
    [1, 2, 1, 0, 2],
    [2, 1, 0, 1, 3],
    [1, 0, 2, 1, 0]
], dtype=np.float32).reshape(1,5,5,1)  # batch size=1

# 🔹 Filter / Kernel (3x3)
K = np.array([
    [1, 0, -1],
    [1, 0, -1],
    [1, 0, -1]
], dtype=np.float32).reshape(3,3,1,1)  # (height, width, in_channels, out_channels)

# 🔹 CNN Model তৈরি
model = Sequential()
conv_layer = Conv2D(
    filters=1,          # output feature map সংখ্যা
    kernel_size=(3,3),  # kernel / filter size
    strides=(1,1),      # stride = 1
    padding='valid',    # padding = 0
    use_bias=False,     # bias = 0, manual calculation এর মতো
    input_shape=(5,5,1)
)
model.add(conv_layer)

# 🔹 ওজন (weights) manually সেট করা
model.layers[0].set_weights([K])  # kernel weights সেট করা

# 🔹 Forward pass — ফিচার ম্যাপ বের করা
feature_map = model.predict(I)
print("Feature Map:\n", feature_map[0,:,:,0])
```
