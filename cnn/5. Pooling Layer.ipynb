```python
# Code 1: Max Pooling 2×2
from tensorflow.keras import layers, models

model = models.Sequential([
    layers.Input(shape=(4,4,1)),
    layers.MaxPooling2D(pool_size=(2,2))
])
model.summary()
```

```python
# Code 2: Average Pooling 2×2
model = models.Sequential([
    layers.Input(shape=(4,4,1)),
    layers.AveragePooling2D(pool_size=(2,2))
])
model.summary()
```

```python
# Code 3: Max Pooling with Stride
model = models.Sequential([
    layers.Input(shape=(6,6,1)),
    layers.MaxPooling2D(pool_size=(2,2), strides=2)
])
model.summary()
```

```python
# Code 4: Average Pooling with Stride
model = models.Sequential([
    layers.Input(shape=(6,6,1)),
    layers.AveragePooling2D(pool_size=(3,3), strides=3)
])
model.summary()
```

```python
# Code 5: Max Pooling + Conv Layer
model = models.Sequential([
    layers.Input(shape=(32,32,3)),
    layers.Conv2D(32, (3,3), activation='relu'),
    layers.MaxPooling2D(2,2)
])
model.summary()
```

```python
# Code 6: Average Pooling + Conv Layer
model = models.Sequential([
    layers.Input(shape=(32,32,3)),
    layers.Conv2D(32, (3,3), activation='relu'),
    layers.AveragePooling2D(2,2)
])
model.summary()
```

```python
# Code 7: Global Max Pooling
model = models.Sequential([
    layers.Input(shape=(8,8,32)),
    layers.GlobalMaxPooling2D()
])
model.summary()
```

```python
# Code 8: Global Average Pooling
model = models.Sequential([
    layers.Input(shape=(8,8,32)),
    layers.GlobalAveragePooling2D()
])
model.summary()
```

```python
# Code 9: Max Pooling + Flatten + Dense
model = models.Sequential([
    layers.Input(shape=(28,28,1)),
    layers.MaxPooling2D(2,2),
    layers.Flatten(),
    layers.Dense(10, activation='softmax')
])
model.summary()
```

```python
# Code 10: Full CNN Example with Pooling
model = models.Sequential([
    layers.Input(shape=(32,32,3)),
    layers.Conv2D(32, (3,3), activation='relu'),
    layers.MaxPooling2D(2,2),
    layers.Conv2D(64, (3,3), activation='relu'),
    layers.AveragePooling2D(2,2),
    layers.Flatten(),
    layers.Dense(128, activation='relu'),
    layers.Dense(10, activation='softmax')
])
model.summary()
```
