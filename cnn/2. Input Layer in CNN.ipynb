```python
from tensorflow.keras import layers, models
# প্রয়োজনীয় লাইব্রেরি ইমপোর্ট করা হচ্ছে

model = models.Sequential([
    layers.Input(shape=(32, 32, 3)),        # ইনপুট লেয়ার (32x32 RGB ইমেজ)
    layers.Rescaling(1./255),               # পিক্সেল মানকে 0–1 রেঞ্জে আনা
    layers.Conv2D(32, (3,3), activation='relu'), # ৩২টি ৩×৩ ফিল্টার প্রয়োগ করা হচ্ছে
    layers.MaxPooling2D(2,2),               # ফিচার ম্যাপ ২×২ করে কমানো হচ্ছে
    layers.Flatten(),                       # সব ফিচার একত্র করে এক ভেক্টর করা হচ্ছে
    layers.Dense(64, activation='relu'),    # ৬৪ নিউরনের হিডেন লেয়ার
    layers.Dense(10, activation='softmax')  # ১০টি ক্লাসে শ্রেণীবিন্যাস
])
```

```python
# Example 1: Basic Input Layer
from tensorflow.keras import layers, models

model = models.Sequential([
    layers.Input(shape=(28, 28, 1))  # ইনপুট লেয়ার → 28x28 গ্রেস্কেল ইমেজ নিচ্ছে
])
model.summary()  # মডেলের গঠন দেখা যাবে
```

```python
# Example 2: Input + Rescaling (Normalization)
from tensorflow.keras import layers, models

model = models.Sequential([
    layers.Input(shape=(28, 28, 1)),  # ইনপুট ইমেজ আকার
    layers.Rescaling(1./255)          # পিক্সেল মান 0–255 থেকে 0–1 করা হচ্ছে
])
model.summary()
```

```python
# Example 3: Input + Conv2D (Feature Extraction শুরু)
from tensorflow.keras import layers, models

model = models.Sequential([
    layers.Input(shape=(64, 64, 3)),          # RGB ইমেজ ইনপুট (64x64)
    layers.Rescaling(1./255),                 # নরমালাইজেশন
    layers.Conv2D(32, (3,3), activation='relu')  # ৩২টি ফিল্টার দিয়ে ফিচার এক্সট্রাকশন
])
model.summary()
```

```python
# Example 4: Input + Flatten + Dense
from tensorflow.keras import layers, models

model = models.Sequential([
    layers.Input(shape=(32, 32, 3)),     # ইনপুট লেয়ার → RGB ইমেজ
    layers.Flatten(),                    # ইনপুটকে 1D তে রূপান্তর করা হচ্ছে
    layers.Dense(64, activation='relu')  # ৬৪ নিউরনের ফুলি কানেক্টেড লেয়ার
])
model.summary()
```

```python
# Example 5: Input + Functional API (Flexible Structure)
from tensorflow.keras import Input, Model
from tensorflow.keras.layers import Conv2D, Flatten, Dense, Rescaling

inputs = Input(shape=(128, 128, 3))             # ইনপুট লেয়ার (128x128 RGB)
x = Rescaling(1./255)(inputs)                   # নরমালাইজেশন লেয়ার
x = Conv2D(16, (3,3), activation='relu')(x)     # কনভলিউশন লেয়ার
x = Flatten()(x)                                # ফ্ল্যাটেন → Dense-এর জন্য প্রস্তুত
outputs = Dense(10, activation='softmax')(x)    # ১০ ক্লাস আউটপুট লেয়ার
model = Model(inputs, outputs)                  # মডেল তৈরি
model.summary()
```

```python
# Example 6: Input + Data Augmentation Layer
from tensorflow.keras import layers, models

data_augmentation = models.Sequential([
    layers.RandomFlip("horizontal"),   # ইমেজ এলোমেলোভাবে উল্টানো
    layers.RandomRotation(0.1)         # সামান্য ঘোরানো (rotation)
])

model = models.Sequential([
    layers.Input(shape=(128, 128, 3)), # ইনপুট লেয়ার
    data_augmentation,                 # ডেটা অগমেন্টেশন প্রয়োগ
    layers.Rescaling(1./255)           # নরমালাইজেশন
])
model.summary()
```

```python
# Example 7: Input + Batch Normalization
from tensorflow.keras import layers, models

model = models.Sequential([
    layers.Input(shape=(32, 32, 3)),        # ইনপুট লেয়ার
    layers.Rescaling(1./255),               # পিক্সেল মান 0–1 এ রূপান্তর
    layers.BatchNormalization(),            # ব্যাচ ভিত্তিক স্কেলিং
    layers.Conv2D(32, (3,3), activation='relu'),  # কনভলিউশন লেয়ার
    layers.MaxPooling2D(2,2),               # ম্যাক্সপুলিং
    layers.Flatten(),                       # ফ্ল্যাটেন
    layers.Dense(10, activation='softmax')   # আউটপুট ক্লাসিফিকেশন
])
model.summary()
```

```python
# Example 8: Input + Dropout Layer
from tensorflow.keras import layers, models

model = models.Sequential([
    layers.Input(shape=(64, 64, 3)),      # ইনপুট
    layers.Rescaling(1./255),             # নরমালাইজেশন
    layers.Flatten(),                     # 3D → 1D
    layers.Dropout(0.5),                  # ৫০% নিউরন এলোমেলোভাবে বাদ দিচ্ছে
    layers.Dense(10, activation='softmax') # আউটপুট লেয়ার
])
model.summary()
```

```python
# Example 9: Input Layer + CNN Architecture (Mini CNN)
from tensorflow.keras import layers, models

model = models.Sequential([
    layers.Input(shape=(32, 32, 3)),              # ইনপুট RGB ইমেজ
    layers.Rescaling(1./255),                     # নরমালাইজেশন
    layers.Conv2D(32, (3,3), activation='relu'),  # প্রথম কনভলিউশন
    layers.MaxPooling2D(2,2),                     # পুলিং
    layers.Conv2D(64, (3,3), activation='relu'),  # দ্বিতীয় কনভলিউশন
    layers.MaxPooling2D(2,2),
    layers.Flatten(),
    layers.Dense(128, activation='relu'),         # Dense layer
    layers.Dense(10, activation='softmax')        # আউটপুট
])
model.summary()
```

```python
# Example 10: Full CNN Training with Input Layer
import tensorflow as tf
from tensorflow.keras import layers, models

# CIFAR-10 ডেটাসেট লোড করা হচ্ছে
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.cifar10.load_data()

# নরমালাইজেশন (ইনপুট মান 0–1)
x_train, x_test = x_train / 255.0, x_test / 255.0

# CNN মডেল তৈরি
model = models.Sequential([
    layers.Input(shape=(32, 32, 3)),             # ইনপুট লেয়ার
    layers.Conv2D(32, (3,3), activation='relu'), # কনভলিউশন লেয়ার
    layers.MaxPooling2D(2,2),
    layers.Flatten(),
    layers.Dense(64, activation='relu'),
    layers.Dense(10, activation='softmax')
])

# মডেল কম্পাইল করা হচ্ছে
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])

# মডেল প্রশিক্ষণ (একটি epoch এর জন্য)
model.fit(x_train, y_train, epochs=1, batch_size=64)
```
