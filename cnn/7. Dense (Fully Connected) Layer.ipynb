```python
# Code 1: Basic Dense Layer
from tensorflow.keras import models, layers

model = models.Sequential([
    layers.Dense(10, input_shape=(5,), activation='relu')  # ৫ ইনপুট, ১০ আউটপুট, relu activation
])
model.summary()
```

```python
# Code 2: Multiple Dense Layers
model = models.Sequential([
    layers.Dense(64, activation='relu'),   # প্রথম Dense লেয়ার
    layers.Dense(32, activation='relu'),   # দ্বিতীয় Dense লেয়ার
    layers.Dense(10, activation='softmax') # আউটপুট লেয়ার
])
model.summary()
```

```python
# Code 3: Dense after Flatten
model = models.Sequential([
    layers.Input(shape=(28, 28, 1)),  # ইনপুট ইমেজ
    layers.Flatten(),                 # ফ্ল্যাটেন
    layers.Dense(128, activation='relu'),
    layers.Dense(10, activation='softmax')
])
model.summary()
```

```python
# Code 4: Custom Dense Example
import tensorflow as tf
x = tf.constant([[1.0, 2.0, 3.0]])          # ইনপুট
dense = tf.keras.layers.Dense(2, activation='relu')  # ২ আউটপুট নিউরন
y = dense(x)
print(y)
```

```python
# Code 5: Sigmoid Activation
model = models.Sequential([
    layers.Dense(1, input_shape=(3,), activation='sigmoid')  # বাইনারি ক্লাসিফিকেশন
])
model.summary()
```

```python
# Code 6: Softmax Output
model = models.Sequential([
    layers.Dense(3, activation='softmax')  # ৩ ক্লাস আউটপুট
])
model.summary()
```

```python
# Code 7: Dense with Bias Disabled
model = models.Sequential([
    layers.Dense(4, use_bias=False)  # বায়াস ছাড়া Dense লেয়ার
])
model.summary()
```

```python
# Code 8: Access Weights
dense = layers.Dense(2)
x = tf.ones((1, 3))
y = dense(x)
print(dense.get_weights())  # ওজন ও বায়াস প্রিন্ট করবে
```

```python
# Code 9: Batch Input
x = tf.random.normal((5, 4))  # ৫টি ইনপুট, প্রতিটি ৪ ফিচার
dense = layers.Dense(3, activation='relu')
y = dense(x)
print(y.shape)  # (5, 3)
```

```python
# Code 10: Model Summary
model = models.Sequential([
    layers.Dense(128, input_shape=(64,), activation='relu'),
    layers.Dense(10, activation='softmax')
])
model.summary()  # Dense লেয়ারের প্যারামিটার সংখ্যা দেখাবে
```

# কেন আমরা একাধিক Dense Layer (Multiple Dense Layers) ব্যবহার করি?

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

model = Sequential([
    Dense(128, activation='relu', input_shape=(784,)),  # প্রথম Dense layer -> সহজ ফিচার শেখে 🧩
    Dense(64, activation='relu'),                       # দ্বিতীয় Dense layer -> গভীর ফিচার শেখে 🧠
    Dense(10, activation='softmax')                     # আউটপুট layer -> 10 ক্লাসে শ্রেণিবিন্যাস 🎯
])
model.summary()
```
