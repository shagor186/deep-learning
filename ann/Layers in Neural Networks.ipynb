# 1. Input Layer

```python
from tensorflow import keras
from tensorflow.keras import layers

# একটি সাধারণ ANN মডেল যেখানে ইনপুট লেয়ার নির্ধারণ করা হয়েছে
model = keras.Sequential([
    layers.Input(shape=(3,)),       # ইনপুট লেয়ারে ৩টি ফিচার
    layers.Dense(4, activation='relu'),  # হিডেন লেয়ার
    layers.Dense(1, activation='sigmoid')  # আউটপুট লেয়ার
])

model.summary()
```

# 2. Hidden Layer

```python
from tensorflow import keras
from tensorflow.keras import layers

model = keras.Sequential([
    layers.Input(shape=(3,)),               # Input Layer (3 ফিচার)
    layers.Dense(4, activation='relu'),     # Hidden Layer-1 (4 নিউরন)
    layers.Dense(3, activation='relu'),     # Hidden Layer-2 (3 নিউরন)
    layers.Dense(1, activation='sigmoid')   # Output Layer
])

model.summary()
```

# 3. Output Layer

```python
# Binary Classification 
from tensorflow import keras
from tensorflow.keras import layers

model = keras.Sequential([
    layers.Input(shape=(3,)),              # Input Layer
    layers.Dense(4, activation='relu'),    # Hidden Layer
    layers.Dense(1, activation='sigmoid')  # Output Layer (Binary Classification)
])

model.summary()
```

```python
# Multiclass Classification 
model = keras.Sequential([
    layers.Input(shape=(3,)),
    layers.Dense(4, activation='relu'),
    layers.Dense(3, activation='softmax')  # 3 classes
])

model.summary()
```

```python
# Regression
model = keras.Sequential([
    layers.Input(shape=(3,)),
    layers.Dense(4, activation='relu'),
    layers.Dense(1, activation='linear')  # Continuous output
])

model.summary()
```
