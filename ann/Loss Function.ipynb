# 1. Regression Loss Function

### 1.1 Mean Squared Error (MSE)

```python
import numpy as np

# True and Predicted values
y_true = np.array([2, 4, 6])
y_pred = np.array([3, 5, 5])

# MSE Calculation
mse = np.mean((y_true - y_pred) ** 2)
print("Mean Squared Error:", mse)
```

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import numpy as np
import matplotlib.pyplot as plt

# Step 1: Data Preparation
X = np.linspace(0, 10, 100)
y = 3 * X + 2 + np.random.randn(100) * 2   # Linear + noise

X = X.reshape(-1, 1)
y = y.reshape(-1, 1)

# Step 2: Model Architecture
model = keras.Sequential([
    layers.Dense(8, activation='relu', input_shape=(1,)),
    layers.Dense(1)  # Linear output layer
])

# Step 3: Compile (MSE as Loss)
model.compile(optimizer='adam', loss='mean_squared_error', metrics=['mae'])

# Step 4: Train
history = model.fit(X, y, epochs=200, verbose=0)

# Step 5: Evaluate
loss, mae = model.evaluate(X, y, verbose=0)
print(f"Final MSE Loss: {loss:.4f}")
print(f"Final MAE: {mae:.4f}")

# Step 6: Predict & Visualize
y_pred = model.predict(X)

plt.figure(figsize=(8,5))
plt.scatter(X, y, color='blue', label='True Data')
plt.plot(X, y_pred, color='red', linewidth=2, label='Model Prediction')
plt.title("Regression using ANN (Loss: Mean Squared Error)")
plt.xlabel("X")
plt.ylabel("y / ŷ")
plt.legend()
plt.show()
```

### 1.2 Mean Absolute Error (MAE)

```python
import numpy as np

# True & Predicted
y_true = np.array([2, 4, 6])
y_pred = np.array([3, 5, 5])

mae = np.mean(np.abs(y_true - y_pred))
print("Mean Absolute Error:", mae)
```

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import numpy as np
import matplotlib.pyplot as plt

# Step 1: Data Preparation
X = np.linspace(0, 10, 100)
y = 3 * X + 2 + np.random.randn(100) * 2

X = X.reshape(-1, 1)
y = y.reshape(-1, 1)

# Step 2: Model Architecture
model = keras.Sequential([
    layers.Dense(8, activation='relu', input_shape=(1,)),
    layers.Dense(1)
])

# Step 3: Compile (Loss = MAE)
model.compile(optimizer='adam', loss='mean_absolute_error', metrics=['mse'])

# Step 4: Train
history = model.fit(X, y, epochs=200, verbose=0)

# Step 5: Evaluate
loss, mse = model.evaluate(X, y, verbose=0)
print(f"Final MAE Loss: {loss:.4f}")
print(f"Final MSE: {mse:.4f}")

# Step 6: Predict & Visualize
y_pred = model.predict(X)

plt.figure(figsize=(8,5))
plt.scatter(X, y, color='blue', label='True Data')
plt.plot(X, y_pred, color='red', linewidth=2, label='Model Prediction')
plt.title("Regression using ANN (Loss: Mean Absolute Error)")
plt.xlabel("X")
plt.ylabel("y / ŷ")
plt.legend()
plt.show()
```
