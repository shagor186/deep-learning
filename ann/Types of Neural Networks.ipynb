# 1. Feedforward Neural Networks (FNN)

```python
import tensorflow as tf
from tensorflow.keras import layers, models

# Feedforward Neural Network Model
model = models.Sequential([
    layers.Dense(8, input_shape=(4,), activation='relu'),  # Hidden Layer
    layers.Dense(4, activation='relu'),
    layers.Dense(1, activation='sigmoid')  # Output Layer
])

# Compile the model
model.compile(optimizer='adam',
              loss='binary_crossentropy',
              metrics=['accuracy'])

# Dummy Training Data
import numpy as np
X = np.random.rand(100, 4)
y = np.random.randint(0, 2, (100,))

# Train the model
model.fit(X, y, epochs=20, batch_size=8, verbose=1)

# Evaluate
loss, acc = model.evaluate(X, y)
print(f"Loss: {loss:.4f}, Accuracy: {acc:.4f}")
```

# 2. Convolutional Neural Networks (CNNs)

```python
import tensorflow as tf
from tensorflow.keras import layers, models

# CNN Model
model = models.Sequential([
    layers.Conv2D(32, (3, 3), activation='relu', input_shape=(28, 28, 1)),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),
    layers.Flatten(),
    layers.Dense(64, activation='relu'),
    layers.Dense(10, activation='softmax')  # 10 classes (e.g., digits)
])

# Compile Model
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

# Example: MNIST dataset
(X_train, y_train), (X_test, y_test) = tf.keras.datasets.mnist.load_data()
X_train = X_train.reshape(-1, 28, 28, 1) / 255.0
X_test = X_test.reshape(-1, 28, 28, 1) / 255.0

# Train
model.fit(X_train, y_train, epochs=5, batch_size=64, validation_split=0.1)

# Evaluate
loss, acc = model.evaluate(X_test, y_test)
print(f"Test Accuracy: {acc:.4f}")
```

# 3. Recurrent Neural Networks (RNNs)

```python
import tensorflow as tf
from tensorflow.keras import layers, models

# Simple RNN Model for Sequential Data
model = models.Sequential([
    layers.SimpleRNN(64, input_shape=(10, 8), activation='tanh'),
    layers.Dense(1, activation='sigmoid')
])

# Compile
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

# Dummy Sequential Data
import numpy as np
X = np.random.rand(100, 10, 8)  # 100 samples, 10 timesteps, 8 features
y = np.random.randint(0, 2, (100,))

# Train
model.fit(X, y, epochs=10, batch_size=8, verbose=1)

# Evaluate
loss, acc = model.evaluate(X, y)
print(f"Accuracy: {acc:.4f}")
```

# 4. Long Short-Term Memory (LSTM)

```python
import tensorflow as tf
from tensorflow.keras import layers, models
import numpy as np

# LSTM Model
model = models.Sequential([
    layers.LSTM(64, input_shape=(10, 8), activation='tanh', return_sequences=False),
    layers.Dense(1, activation='sigmoid')
])

# Compile
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

# Dummy Sequential Data
X = np.random.rand(100, 10, 8)  # 100 samples, 10 timesteps, 8 features
y = np.random.randint(0, 2, (100,))

# Train
model.fit(X, y, epochs=10, batch_size=8, verbose=1)

# Evaluate
loss, acc = model.evaluate(X, y)
print(f"LSTM Model Accuracy: {acc:.4f}")
```

# 5. Gated Recurrent Unit (GRU)

```python
import tensorflow as tf
from tensorflow.keras import layers, models
import numpy as np

# GRU Model
model = models.Sequential([
    layers.GRU(64, input_shape=(10, 8), activation='tanh', return_sequences=False),
    layers.Dense(1, activation='sigmoid')
])

# Compile Model
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

# Dummy Sequential Data
X = np.random.rand(100, 10, 8)
y = np.random.randint(0, 2, (100,))

# Train
model.fit(X, y, epochs=10, batch_size=8, verbose=1)

# Evaluate
loss, acc = model.evaluate(X, y)
print(f"GRU Model Accuracy: {acc:.4f}")
```

# 6. Generative Adversarial Networks (GANs)

```python
import tensorflow as tf
from tensorflow.keras import layers
import numpy as np

# Generator
def build_generator(latent_dim):
    model = tf.keras.Sequential([
        layers.Dense(128, activation='relu', input_dim=latent_dim),
        layers.Dense(784, activation='sigmoid')  # Output: 28x28 flattened image
    ])
    return model

# Discriminator
def build_discriminator():
    model = tf.keras.Sequential([
        layers.Dense(128, activation='relu', input_shape=(784,)),
        layers.Dense(1, activation='sigmoid')
    ])
    model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
    return model

# GAN Model
latent_dim = 100
generator = build_generator(latent_dim)
discriminator = build_discriminator()
discriminator.trainable = False

gan_input = layers.Input(shape=(latent_dim,))
gan_output = discriminator(generator(gan_input))
gan = tf.keras.Model(gan_input, gan_output)
gan.compile(optimizer='adam', loss='binary_crossentropy')

# Dummy Training Example
batch_size = 16
real_data = np.random.rand(batch_size, 784)
noise = np.random.rand(batch_size, latent_dim)

# Train discriminator
d_loss_real = discriminator.train_on_batch(real_data, np.ones((batch_size,1)))
d_loss_fake = discriminator.train_on_batch(generator.predict(noise), np.zeros((batch_size,1)))

# Train generator via GAN
g_loss = gan.train_on_batch(noise, np.ones((batch_size,1)))

print(f"D_loss_real: {d_loss_real}, D_loss_fake: {d_loss_fake}, G_loss: {g_loss}")
```
